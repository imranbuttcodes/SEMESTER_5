## Exercise 2.1. 

```python
def selection_sort(items):
    for steps in range(len(items)):
        location_of_smallest = steps
        for location in range(steps + 1, len(items)):
                if items[location] < items[location_of_smallest]:
                    location_of_smallest = location
        items[steps], items[location_of_smallest] = items[location_of_smallest], items[steps]
    return items

lst = [4,2,3,1,5]

print(selection_sort(lst))
```

## Exercise 2.2.

```python
def merge(left, right):
    i = j = 0

    result = []
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result
```

```python
def merge_sort(arr):

    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)
```

```python
lst = [4, 2, 3, 1, 5]
print(merge_sort(lst))
```

```python
def insertion_sort(arr):
    for i in range(1, len(arr) - 1):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j] 
            j = j - 1

        arr[j + 1] = key

    return arr
```

```python
lst = [4, 2, 3, 1, 5]
print(insertion_sort(lst))
```

## Exercise 2.4.

```python

def statistics(numbers):
    
    total = sum(numbers)
    mean = total / len(numbers)

    
    numbers.sort()
    n = len(numbers)

    if n % 2 == 1:
        median = numbers[n // 2]
    else:
        median = (numbers[n // 2 - 1] + numbers[n // 2]) / 2

    mode = max(numbers, key=numbers.count)

    print("Sum:", total)
    print("Mean:", mean)
    print("Median:", median)
    print("Mode:", mode)
    


numbers = [1,2,5,4,3]

statistics(numbers)
```

## Exercise 2.5

```python
def dups(numbers):
    duplicates = []

    for num in numbers:
        if numbers.count(num) > 1 and num not in duplicates:
            duplicates.append(num)

    return duplicates


numbers = [1,4,3,2,2,4]
result = dups(numbers)

print("Duplicate elements:", result)
```

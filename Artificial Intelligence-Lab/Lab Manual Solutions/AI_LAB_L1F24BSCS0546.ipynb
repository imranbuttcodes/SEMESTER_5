```python
def sundaes():
    flavors = ["vanilla", "chocolate", "strawberry", "pistacchio"]
    sauces = ["caramel", "butterscotch", "chocolate"]
    total = 0
    for flavor in flavors:
        for sauce in sauces:
            print(flavor + " ice cream sundae with " + sauce + " sauce")
            total += 1

    return total
```

```python
def triangle():
    value = 1
    row = 1
    while row <= 10:
        column = 1
        while column <= row:
            if column != row:
                print(value, ' ', sep = '', end = '')
            else:
                print(value)
            value = value + 1
            column = column + 1
        row = row + 1
```

```python
def factorial(n):
    if n == 0:
        return 1

    return n * factorial(n-1)
```

```python
def count_pattern(pattern, lst):
    
    total = 0
    
    for i in range(len(lst)-len(pattern) + 1):
        found = True
        for j in range(len(pattern)):
            if (pattern[j] != lst[i+j]):
                found = False
                break
        if found:
            total += 1        

    return total
```

```python
print(count_pattern(('a', 'b', 'a'), ('g', 'a', 'b', 'a', 'b',
'a','b', 'a')))
```

```python
def calculator(fst, scd, opt):
    if opt == "+":
        return fst * scd
    elif opt == "-":
        return fst - scd
    elif opt == "*":
        return fst * scd
    elif opt == "/":
        if scd == 0:
            return "Can't devide a number by 0"
        return fst / scd
```

```python
print(calculator(4,2,"/"))
```

```python
class RomanNumeral:
    def __init__(self, number):
        self.number = number

    def convert(self):
        values = [
            (1000, "M"),
            (900, "CM"),
            (500, "D"),
            (400, "CD"),
            (100, "C"),
            (90, "XC"),
            (50, "L"),
            (40, "XL"),
            (10, "X"),
            (9, "IX"),
            (5, "V"),
            (4, "IV"),
            (1, "I")
        ]

        roman = ""

        for value, symbol in values:
            while self.number >= value:
                roman += symbol
                self.number -= value

        return roman


```

```python
number = int(input("Enter an integer: "))
print(f"Number Entered: {number}")
roman_number = RomanNumeral(number)

print("Roman numeral:", roman_number.convert())
```

```python
class PairElements:
    def __init__(self, array):
        self.array = array.sort()

    def find_pair(self, target):
        st = 0
        end = len(self.array) - 1

        while st <= end:
            if self.array[st] + self.array[end] == target:
                return st, end
            elif self.array[st] + self.array[end] > target:
                end -= 1
            else:
                st += 1
        return "pair Not Found"
```

```python
numbers= [10,20,10,40,50,60,70]

pair_elements = PairElements(numbers)

print(pair_elements.find_pair(50))
```

```python
class ThreeSum:

    def __init__(self, array):
        self.array = array

    def find_triplets(self):

        self.array.sort()
        result = []

        for i in range(len(self.array) - 2):

            left = i + 1
            right = len(self.array) - 1

            while left < right:

                total = self.array[i] + self.array[left] + self.array[right]

                if total == 0:
                    result.append([
                        self.array[i],
                        self.array[left],
                        self.array[right]
                    ])
                    left += 1
                    right -= 1

                elif total < 0:
                    left += 1

                else:
                    right -= 1

        return result


obj = ThreeSum([-25, -10, -7, -3, 2, 4, 8, 10])

print(obj.find_triplets())
```

```python
def reverse_str(str):
    if len(str) <= 1:
        return str
    return reverse_str(str[1:]) + str[0]
```

```python
class Numbers:

    MULTIPLIER = 2

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def add(self):
        return self.x + self.y

    
    @classmethod
    def multiply(cls, a):
        return a * cls.MULTIPLIER

    @staticmethod
    def subtract(b, c):
        return b - c

    @property
    def value(self):
        return self.x, self.y

    
    @value.setter
    def value(self, values):
        self.x, self.y = values

    @value.deleter
    def value(self):
        del self.x
        del self.y



num = Numbers(10, 5)

print("Addition:", num.add())
print("Multiplication:", Numbers.multiply(5))
print("Subtraction:", Numbers.subtract(10, 5))
print("Values:", num.value)


num.value = (20, 30)
print("New values:", num.value)


del num.value
```

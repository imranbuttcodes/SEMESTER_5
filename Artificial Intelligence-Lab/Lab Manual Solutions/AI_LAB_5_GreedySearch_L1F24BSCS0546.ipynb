## Task 1

```python
def greedy_search(graph, heuristic, start, goal):
    open_list = [(heuristic[start], start)]
    visited = set()
    parent = {start:None}

    while open_list:
        open_list.sort()

        h, curr = open_list.pop(0)

        if curr in visited:
            continue

        visited.add(curr)
        
        if curr == goal:
            path = []

            node = goal

            while node is not None:
                path.append(node)
                node = parent[node]

            path.reverse()

            return path
        


        for child in graph[curr]:
            if child not in visited and child not in parent:
                open_list.append((heuristic[child], child))
                parent[child] = curr
        

    print("Goal Not Found")
    
    return None

```

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

# Heuristic values
heuristic = {
    'A': 6,
    'B': 4,
    'C': 2,
    'D': 5,
    'E': 3,
    'F': 1,
    'G': 0
}

print("Path:",greedy_search(graph, heuristic, 'A', 'G'))
```

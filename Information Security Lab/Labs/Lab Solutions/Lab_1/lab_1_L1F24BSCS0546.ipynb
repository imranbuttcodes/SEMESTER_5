## Caesar Cipher


```python
def caesar_cipher_encrypt(ciphertext, shift_key):
    ecnrypted_text = []

    for char in ciphertext:
        if char.isupper():
            encrypted_char = chr(((ord(char) - ord('A') + shift_key) % 26) + ord('A'))
            ecnrypted_text.append(encrypted_char)
        elif char.islower():
            encrypted_char = chr(((ord(char) - ord('a') + shift_key) % 26) + ord('a'))
            ecnrypted_text.append(encrypted_char)
        else:
            ecnrypted_text.append(char)

    return "".join(ecnrypted_text)
```

```python
print(caesar_cipher_encrypt("IMRAN", 3))
```

```python
def caesar_cipher_decrypt(ciphertext, shift_key):
    decrypted_text = []

    for char in ciphertext:
        if char.isupper():
            decrypted_char = chr(((ord(char) - ord('A') - shift_key) % 26) + ord('A'))
            decrypted_text.append(decrypted_char)
        elif char.islower():
            decrypted_char = chr(((ord(char) - ord('a') - shift_key) % 26) + ord('a'))
            decrypted_text.append(decrypted_char)
        else:
            decrypted_text.append(char)

    return "".join(decrypted_text)
```

```python
print(caesar_cipher_decrypt("LPUDQ", 3))
```

```python
def vigenère_cipher_encrypt(plan_text, key):
    new_key = key
    for i in range(len(plan_text)-len(key)):
        new_key += key[i%len(key)]

    encrypted_text = ""

    for p, k in zip(plan_text,new_key):
        if p.isupper():
            encrypted_text += chr((((ord(p) - ord('A')) + (ord(k) - ord('A'))) % 26 ) + ord('A'))
        elif p.islower():
            encrypted_text += chr((((ord(p) - ord('a')) + (ord(k) - ord('a'))) % 26 ) + ord('a'))
    return encrypted_text
```

```python
print(vigenère_cipher_encrypt("DEFENDTHEEASTWALL", "KEY"))
```

```python
def vigenère_cipher_decrypt(plan_text, key):
    new_key = key
    for i in range(len(plan_text)-len(key)):
        new_key += key[i%len(key)]

    decrypted_text = ""

    for p, k in zip(plan_text,new_key):
        if p.isupper():
            decrypted_text += chr((((ord(p) - ord('A')) - (ord(k) - ord('A'))) % 26 ) + ord('A'))
        elif p.islower():
            decrypted_text += chr((((ord(p) - ord('a')) - (ord(k) - ord('a'))) % 26 ) + ord('a'))
    return decrypted_text
```

```python
print(vigenère_cipher_decrypt("GIKLFOZHAKDSOYHIV", "KEY"))
```

```python
def auto_key_cipher_encrypt(plan_text, key):
    new_key = key + plan_text[:len(plan_text) - len(key)]

    encrypted_text = ""
    for p, k in zip(plan_text,new_key):
            if p.isupper():
                encrypted_text += chr((((ord(p) - ord('A')) + (ord(k) - ord('A'))) % 26 ) + ord('A'))
            elif p.islower():
                encrypted_text += chr((((ord(p) - ord('a')) + (ord(k) - ord('a'))) % 26 ) + ord('a'))
    return encrypted_text       
```

```python
auto_key_cipher_encrypt('CRYPTOGRAPHY', 'KEY')
```

```python
def auto_key_cipher_decrypt(cipher_text, key):
    decrypted_text = ""

    for i, c in enumerate(cipher_text):
        if i < len(key):
            k = key[i]
        else:
            k = decrypted_text[i - len(key)]

        if c.isupper():
            decrypted_text += chr(
                ((ord(c) - ord('A')) - (ord(k) - ord('A'))) % 26
                + ord('A')
            )

        elif c.islower():
            decrypted_text += chr(
                ((ord(c) - ord('a')) - (ord(k) - ord('a'))) % 26
                + ord('a')
            )

    return decrypted_text
```
